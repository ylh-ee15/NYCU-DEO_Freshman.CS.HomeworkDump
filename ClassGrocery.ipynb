```python
class Cart:
    def __init__(self):
        self.items = []

    def __str__(self):
        return ", ".join(self.items)

    def add_item(self, items):
        self.items.extend(items.split())

    def remove_item(self, items):
        for item in items.split():
            if item in self.items:
                self.items.remove(item)
            else:
                print(f"{item} 不存在購物車中")

    def checkout(self):
        return self.items


shopping_cart = Cart()
print (f"購物車已建立，目前內容 [{shopping_cart}]")
shopping_cart.add_item(input("輸入您想加入購物車的商品："))
shopping_cart.remove_item(input("輸入您想從購物車刪除的商品："))
final_items = shopping_cart.checkout()
print(f"購物車最終內容：{final_items}\n感謝購買！")
```
